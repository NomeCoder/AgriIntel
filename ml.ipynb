```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("arjunyadav99/indian-agriculture-crop-production-and-yield")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd

csv_file = [f for f in os.listdir(path) if f.endswith(".csv")][0]

df = pd.read_csv(os.path.join(path, csv_file))

print(df.head())
print(df.columns)
```

```python
states = ["Punjab", "Uttar Pradesh", "Maharashtra", "Tamil Nadu"]

filtered_df = df[df["State_Name"].isin(states)]

print(filtered_df.head())
print(filtered_df["State_Name"].value_counts())
```

```python
import pandas as pd

# CSV files
files = [
    "Nutrient (3).csv",   # 2023-24
    "Nutrient (2).csv",   # 2024-25
    "Nutrient (1).csv",   # 2025-26
    "Nutrient.csv"        # 2026-27
]

# States to keep
states = [
    "PUNJAB",
    "MAHARASHTRA",
    "UTTAR PRADESH",
    "TAMIL NADU"
]

data = []

for file in files:

    # Read CSV
    df = pd.read_csv(file, skiprows=2)

    # Clean State column
    df["State"] = (
        df["State"]
        .astype(str)
        .str.strip()
        .str.upper()
    )

    # Extract only required states
    df = df[df["State"].isin(states)]

    # Add to list
    data.append(df)

# Merge all four years
final_df = pd.concat(data, ignore_index=True)

# Save extracted data
final_df.to_csv(
    "soil_selected_states.csv",
    index=False
)

print("Extraction completed!")
print("Total rows:", len(final_df))
print(final_df[["State", "Cycle"]])
```
