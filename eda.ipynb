```python
# ============================================================
# AGRICULTURAL DATASET - COMPLETE EDA
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

sns.set_theme(style="whitegrid")


# -----------------------------
# 2. LOAD DATASET
# -----------------------------

file_path = "agriIntel_merged_final_v.csv.xlsx"

df = pd.read_excel(file_path)

print("Dataset loaded successfully!")


# -----------------------------
# 3. BASIC DATASET INFORMATION
# -----------------------------

print("\n" + "="*60)
print("BASIC DATASET INFORMATION")
print("="*60)

print("Shape:", df.shape)

print("\nNumber of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()


# -----------------------------
# 4. FIRST AND LAST RECORDS
# -----------------------------

print("\n" + "="*60)
print("FIRST 5 RECORDS")
print("="*60)

print(df.head())

print("\n" + "="*60)
print("LAST 5 RECORDS")
print("="*60)

print(df.tail())


# -----------------------------
# 5. RANDOM SAMPLE
# -----------------------------

print("\nRandom sample:")
print(df.sample(10))


# -----------------------------
# 6. UNIQUE VALUES
# -----------------------------

print("\n" + "="*60)
print("UNIQUE VALUES")
print("="*60)

for col in df.columns:
    print(f"{col}: {df[col].nunique()} unique values")


# -----------------------------
# 7. MISSING VALUE ANALYSIS
# -----------------------------

print("\n" + "="*60)
print("MISSING VALUE ANALYSIS")
print("="*60)

missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": df.isnull().mean() * 100
})

missing = missing.sort_values(
    by="Missing_Count",
    ascending=False
)

print(missing)


# Plot missing values
missing_nonzero = missing[missing["Missing_Count"] > 0]

if len(missing_nonzero) > 0:

    plt.figure(figsize=(12, 6))

    sns.barplot(
        x=missing_nonzero.index,
        y=missing_nonzero["Missing_Percentage"]
    )

    plt.xticks(rotation=90)
    plt.ylabel("Missing Percentage (%)")
    plt.title("Missing Values by Column")
    plt.tight_layout()
    plt.show()

else:
    print("No missing values found!")


# -----------------------------
# 8. DUPLICATE ANALYSIS
# -----------------------------

print("\n" + "="*60)
print("DUPLICATE ANALYSIS")
print("="*60)

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

print(
    "Duplicate percentage:",
    round((duplicate_count / len(df)) * 100, 2),
    "%"
)


# -----------------------------
# 9. DESCRIPTIVE STATISTICS
# -----------------------------

print("\n" + "="*60)
print("NUMERICAL DESCRIPTIVE STATISTICS")
print("="*60)

print(df.describe())


print("\n" + "="*60)
print("CATEGORICAL DESCRIPTIVE STATISTICS")
print("="*60)

print(df.describe(include="object"))


# -----------------------------
# 10. SEPARATE NUMERICAL/CATEGORICAL
# -----------------------------

numeric_cols = df.select_dtypes(
    include=np.number
).columns.tolist()

categorical_cols = df.select_dtypes(
    include="object"
).columns.tolist()

print("\nNumerical columns:")
print(numeric_cols)

print("\nCategorical columns:")
print(categorical_cols)


# ============================================================
# 11. CATEGORICAL VARIABLE ANALYSIS
# ============================================================

for col in categorical_cols:

    print("\n" + "="*60)
    print(f"VALUE COUNTS: {col}")
    print("="*60)

    print(df[col].value_counts().head(20))


# ============================================================
# 12. STATE ANALYSIS
# ============================================================

if "State" in df.columns:

    print("\n" + "="*60)
    print("STATE ANALYSIS")
    print("="*60)

    print("Number of states:",
          df["State"].nunique())

    print("\nTop states:")
    print(df["State"].value_counts().head(20))

    plt.figure(figsize=(14, 7))

    df["State"].value_counts().head(20).plot(
        kind="bar"
    )

    plt.title("Top 20 States by Number of Records")
    plt.xlabel("State")
    plt.ylabel("Number of Records")
    plt.xticks(rotation=75)

    plt.tight_layout()
    plt.show()


# ============================================================
# 13. DISTRICT ANALYSIS
# ============================================================

if "District" in df.columns:

    print("\nNumber of districts:",
          df["District"].nunique())

    print("\nTop 20 districts:")
    print(df["District"].value_counts().head(20))

    plt.figure(figsize=(14, 7))

    df["District"].value_counts().head(20).plot(
        kind="bar"
    )

    plt.title("Top 20 Districts by Number of Records")
    plt.xlabel("District")
    plt.ylabel("Number of Records")
    plt.xticks(rotation=75)

    plt.tight_layout()
    plt.show()


# ============================================================
# 14. CROP ANALYSIS
# ============================================================

if "Crop" in df.columns:

    print("\n" + "="*60)
    print("CROP ANALYSIS")
    print("="*60)

    print("Number of crops:",
          df["Crop"].nunique())

    print("\nMost common crops:")
    print(df["Crop"].value_counts().head(20))

    plt.figure(figsize=(14, 7))

    df["Crop"].value_counts().head(20).plot(
        kind="bar"
    )

    plt.title("Top 20 Crops")
    plt.xlabel("Crop")
    plt.ylabel("Number of Records")
    plt.xticks(rotation=75)

    plt.tight_layout()
    plt.show()


# ============================================================
# 15. SEASON ANALYSIS
# ============================================================

if "Season" in df.columns:

    print("\nSeason distribution:")
    print(df["Season"].value_counts())

    plt.figure(figsize=(8, 5))

    sns.countplot(
        data=df,
        x="Season"
    )

    plt.title("Distribution of Agricultural Seasons")
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


# ============================================================
# 16. YEAR ANALYSIS
# ============================================================

if "Year" in df.columns:

    print("\nYear distribution:")
    print(df["Year"].value_counts().sort_index())

    plt.figure(figsize=(14, 6))

    df["Year"].value_counts().sort_index().plot(
        kind="bar"
    )

    plt.title("Number of Records by Year")
    plt.xlabel("Year")
    plt.ylabel("Records")

    plt.xticks(rotation=75)

    plt.tight_layout()
    plt.show()


# ============================================================
# 17. NUMERICAL DISTRIBUTIONS
# ============================================================

for col in numeric_cols:

    plt.figure(figsize=(8, 5))

    sns.histplot(
        df[col].dropna(),
        kde=True,
        bins=40
    )

    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()


# ============================================================
# 18. BOX PLOTS - OUTLIER ANALYSIS
# ============================================================

for col in numeric_cols:

    plt.figure(figsize=(10, 4))

    sns.boxplot(
        x=df[col]
    )

    plt.title(f"Boxplot - {col}")

    plt.tight_layout()
    plt.show()


# ============================================================
# 19. OUTLIER COUNT USING IQR
# ============================================================

print("\n" + "="*60)
print("OUTLIER ANALYSIS")
print("="*60)

outlier_summary = []

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ][col].count()

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": outliers,
        "Outlier_Percentage":
            (outliers / len(df)) * 100
    })

outlier_df = pd.DataFrame(outlier_summary)

print(outlier_df.sort_values(
    "Outlier_Count",
    ascending=False
))


# ============================================================
# 20. CORRELATION MATRIX
# ============================================================

correlation = df[numeric_cols].corr()

print("\n" + "="*60)
print("CORRELATION MATRIX")
print("="*60)

print(correlation)


plt.figure(figsize=(18, 14))

sns.heatmap(
    correlation,
    cmap="coolwarm",
    center=0,
    annot=False
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()


# ============================================================
# 21. STRONG CORRELATIONS
# ============================================================

corr_pairs = (
    correlation
    .where(np.triu(np.ones(correlation.shape), k=1).astype(bool))
    .stack()
    .sort_values(ascending=False)
)

print("\n" + "="*60)
print("STRONGEST POSITIVE CORRELATIONS")
print("="*60)

print(corr_pairs.head(15))


print("\n" + "="*60)
print("STRONGEST NEGATIVE CORRELATIONS")
print("="*60)

print(corr_pairs.tail(15))


# ============================================================
# 22. AREA vs PRODUCTION
# ============================================================

if "Area" in df.columns and "Production" in df.columns:

    plt.figure(figsize=(9, 6))

    sns.scatterplot(
        data=df,
        x="Area",
        y="Production",
        alpha=0.4
    )

    plt.title("Area vs Production")
    plt.xlabel("Area")
    plt.ylabel("Production")

    plt.tight_layout()
    plt.show()


# ============================================================
# 23. AREA vs YIELD
# ============================================================

if "Area" in df.columns and "Yield" in df.columns:

    plt.figure(figsize=(9, 6))

    sns.scatterplot(
        data=df,
        x="Area",
        y="Yield",
        alpha=0.4
    )

    plt.title("Area vs Yield")
    plt.xlabel("Area")
    plt.ylabel("Yield")

    plt.tight_layout()
    plt.show()


# ============================================================
# 24. RAINFALL vs PRODUCTION
# ============================================================

if "Rainfall_mm" in df.columns and "Production" in df.columns:

    plt.figure(figsize=(9, 6))

    sns.scatterplot(
        data=df,
        x="Rainfall_mm",
        y="Production",
        alpha=0.4
    )

    plt.title("Rainfall vs Production")

    plt.tight_layout()
    plt.show()


# ============================================================
# 25. RAINFALL vs YIELD
# ============================================================

if "Rainfall_mm" in df.columns and "Yield" in df.columns:

    plt.figure(figsize=(9, 6))

    sns.scatterplot(
        data=df,
        x="Rainfall_mm",
        y="Yield",
        alpha=0.4
    )

    plt.title("Rainfall vs Yield")

    plt.tight_layout()
    plt.show()


# ============================================================
# 26. IRRIGATION ANALYSIS
# ============================================================

irrigation_cols = [
    "total_irrigated_ha",
    "canal_irrigated_ha",
    "well_irrigated_ha",
    "tank_irrigated_ha",
    "other_irrigated_ha",
    "irrigation_intensity"
]

existing_irrigation = [
    col for col in irrigation_cols
    if col in df.columns
]

if existing_irrigation:

    print("\n" + "="*60)
    print("IRRIGATION STATISTICS")
    print("="*60)

    print(
        df[existing_irrigation].describe()
    )


# ============================================================
# 27. SOIL ANALYSIS
# ============================================================

soil_cols = [
    "Nitrogen_index",
    "Phosphorus_index",
    "Potassium_index",
    "Organic_Carbon_index",
    "pct_acidic",
    "pct_alkaline",
    "pct_saline"
]

existing_soil = [
    col for col in soil_cols
    if col in df.columns
]

if existing_soil:

    print("\n" + "="*60)
    print("SOIL STATISTICS")
    print("="*60)

    print(
        df[existing_soil].describe()
    )


# ============================================================
# 28. RAINFALL ANALYSIS
# ============================================================

rain_cols = [
    "Rainfall_mm",
    "Mean_Daily_Rainfall_mm",
    "Rainy_Days",
    "Heavy_Rain_Days",
    "Max_Daily_Rainfall_mm"
]

existing_rain = [
    col for col in rain_cols
    if col in df.columns
]

if existing_rain:

    print("\n" + "="*60)
    print("RAINFALL STATISTICS")
    print("="*60)

    print(
        df[existing_rain].describe()
    )


# ============================================================
# 29. PRODUCTION ANALYSIS
# ============================================================

if "Production" in df.columns:

    print("\n" + "="*60)
    print("PRODUCTION ANALYSIS")
    print("="*60)

    print(df["Production"].describe())

    print("\nHighest production records:")

    print(
        df.sort_values(
            "Production",
            ascending=False
        ).head(10)[
            ["State", "District", "Crop", "Year", "Production"]
        ]
    )


# ============================================================
# 30. YIELD ANALYSIS
# ============================================================

if "Yield" in df.columns:

    print("\n" + "="*60)
    print("YIELD ANALYSIS")
    print("="*60)

    print(df["Yield"].describe())

    print("\nHighest yield records:")

    print(
        df.sort_values(
            "Yield",
            ascending=False
        ).head(10)[
            ["State", "District", "Crop", "Year", "Yield"]
        ]
    )


# ============================================================
# 31. TOP CROPS BY TOTAL PRODUCTION
# ============================================================

if "Crop" in df.columns and "Production" in df.columns:

    crop_production = (
        df.groupby("Crop")["Production"]
        .sum()
        .sort_values(ascending=False)
        .head(20)
    )

    print("\nTop crops by total production:")
    print(crop_production)

    plt.figure(figsize=(14, 7))

    crop_production.plot(
        kind="bar"
    )

    plt.title("Top 20 Crops by Total Production")
    plt.xlabel("Crop")
    plt.ylabel("Total Production")

    plt.xticks(rotation=75)

    plt.tight_layout()
    plt.show()


# ============================================================
# 32. TOP STATES BY PRODUCTION
# ============================================================

if "State" in df.columns and "Production" in df.columns:

    state_production = (
        df.groupby("State")["Production"]
        .sum()
        .sort_values(ascending=False)
        .head(20)
    )

    print("\nTop states by production:")
    print(state_production)

    plt.figure(figsize=(14, 7))

    state_production.plot(
        kind="bar"
    )

    plt.title("Top 20 States by Total Production")

    plt.xticks(rotation=75)

    plt.tight_layout()
    plt.show()


# ============================================================
# 33. AVERAGE YIELD BY CROP
# ============================================================

if "Crop" in df.columns and "Yield" in df.columns:

    crop_yield = (
        df.groupby("Crop")["Yield"]
        .mean()
        .sort_values(ascending=False)
        .head(20)
    )

    print("\nTop crops by average yield:")
    print(crop_yield)

    plt.figure(figsize=(14, 7))

    crop_yield.plot(
        kind="bar"
    )

    plt.title("Top 20 Crops by Average Yield")

    plt.xticks(rotation=75)

    plt.tight_layout()
    plt.show()


# ============================================================
# 34. SEASON vs PRODUCTION
# ============================================================

if (
    "Season" in df.columns
    and "Production" in df.columns
):

    season_production = (
        df.groupby("Season")["Production"]
        .sum()
        .sort_values(ascending=False)
    )

    print("\nProduction by season:")
    print(season_production)

    plt.figure(figsize=(9, 5))

    sns.barplot(
        x=season_production.index,
        y=season_production.values
    )

    plt.title("Total Production by Season")

    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


# ============================================================
# 35. SEASON vs YIELD
# ============================================================

if (
    "Season" in df.columns
    and "Yield" in df.columns
):

    season_yield = (
        df.groupby("Season")["Yield"]
        .mean()
        .sort_values(ascending=False)
    )

    print("\nAverage yield by season:")
    print(season_yield)

    plt.figure(figsize=(9, 5))

    sns.barplot(
        x=season_yield.index,
        y=season_yield.values
    )

    plt.title("Average Yield by Season")

    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


# ============================================================
# 36. YEAR-WISE PRODUCTION
# ============================================================

if (
    "Year_start" in df.columns
    and "Production" in df.columns
):

    yearly_production = (
        df.groupby("Year_start")["Production"]
        .sum()
    )

    plt.figure(figsize=(12, 6))

    sns.lineplot(
        x=yearly_production.index,
        y=yearly_production.values,
        marker="o"
    )

    plt.title("Year-wise Total Production")
    plt.xlabel("Year")
    plt.ylabel("Total Production")

    plt.tight_layout()
    plt.show()


# ============================================================
# 37. YEAR-WISE AVERAGE YIELD
# ============================================================

if (
    "Year_start" in df.columns
    and "Yield" in df.columns
):

    yearly_yield = (
        df.groupby("Year_start")["Yield"]
        .mean()
    )

    plt.figure(figsize=(12, 6))

    sns.lineplot(
        x=yearly_yield.index,
        y=yearly_yield.values,
        marker="o"
    )

    plt.title("Year-wise Average Yield")
    plt.xlabel("Year")
    plt.ylabel("Average Yield")

    plt.tight_layout()
    plt.show()


# ============================================================
# 38. IRRIGATION INTENSITY vs YIELD
# ============================================================

if (
    "irrigation_intensity" in df.columns
    and "Yield" in df.columns
):

    plt.figure(figsize=(9, 6))

    sns.scatterplot(
        data=df,
        x="irrigation_intensity",
        y="Yield",
        alpha=0.4
    )

    plt.title(
        "Irrigation Intensity vs Yield"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 39. ORGANIC CARBON vs YIELD
# ============================================================

if (
    "Organic_Carbon_index" in df.columns
    and "Yield" in df.columns
):

    plt.figure(figsize=(9, 6))

    sns.scatterplot(
        data=df,
        x="Organic_Carbon_index",
        y="Yield",
        alpha=0.4
    )

    plt.title(
        "Organic Carbon vs Yield"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 40. NPK vs YIELD
# ============================================================

npk_cols = [
    "Nitrogen_index",
    "Phosphorus_index",
    "Potassium_index"
]

available_npk = [
    col for col in npk_cols
    if col in df.columns
]

if "Yield" in df.columns:

    for col in available_npk:

        plt.figure(figsize=(9, 6))

        sns.scatterplot(
            data=df,
            x=col,
            y="Yield",
            alpha=0.4
        )

        plt.title(
            f"{col} vs Yield"
        )

        plt.tight_layout()
        plt.show()


# ============================================================
# 41. PAIRPLOT FOR IMPORTANT VARIABLES
# ============================================================

important_cols = [
    "Area",
    "Production",
    "Yield",
    "Rainfall_mm",
    "Rainy_Days",
    "irrigation_intensity",
    "Nitrogen_index",
    "Phosphorus_index",
    "Potassium_index",
    "Organic_Carbon_index"
]

important_cols = [
    col for col in important_cols
    if col in df.columns
]

# Use sample to prevent excessive computation
sample_df = df[important_cols].sample(
    min(2000, len(df)),
    random_state=42
)

sns.pairplot(
    sample_df,
    corner=True
)

plt.show()


# ============================================================
# 42. TOP 20 MOST CORRELATED FEATURES WITH YIELD
# ============================================================

if "Yield" in correlation.columns:

    yield_corr = (
        correlation["Yield"]
        .drop("Yield")
        .abs()
        .sort_values(ascending=False)
    )

    print("\n" + "="*60)
    print("FEATURES MOST CORRELATED WITH YIELD")
    print("="*60)

    print(yield_corr.head(20))


# ============================================================
# 43. TOP 20 MOST CORRELATED FEATURES WITH PRODUCTION
# ============================================================

if "Production" in correlation.columns:

    production_corr = (
        correlation["Production"]
        .drop("Production")
        .abs()
        .sort_values(ascending=False)
    )

    print("\n" + "="*60)
    print("FEATURES MOST CORRELATED WITH PRODUCTION")
    print("="*60)

    print(production_corr.head(20))


# ============================================================
# 44. DATA QUALITY CHECK
# ============================================================

print("\n" + "="*60)
print("DATA QUALITY CHECK")
print("="*60)

print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Duplicate rows:", df.duplicated().sum())
print("Missing values:", df.isnull().sum().sum())


# Negative-value check
print("\nNegative values in numerical columns:")

negative_summary = {}

for col in numeric_cols:

    count = (df[col] < 0).sum()

    if count > 0:
        negative_summary[col] = count

print(negative_summary)


# ============================================================
# 45. FINAL EDA SUMMARY
# ============================================================

print("\n" + "="*70)
print("FINAL EDA SUMMARY")
print("="*70)

print(f"""
Dataset Shape:
    Rows    : {df.shape[0]}
    Columns : {df.shape[1]}

Numerical Columns:
    {len(numeric_cols)}

Categorical Columns:
    {len(categorical_cols)}

Duplicate Rows:
    {df.duplicated().sum()}

Total Missing Values:
    {df.isnull().sum().sum()}

Number of States:
    {df['State'].nunique() if 'State' in df.columns else 'N/A'}

Number of Districts:
    {df['District'].nunique() if 'District' in df.columns else 'N/A'}

Number of Crops:
    {df['Crop'].nunique() if 'Crop' in df.columns else 'N/A'}

Number of Seasons:
    {df['Season'].nunique() if 'Season' in df.columns else 'N/A'}
""")
```
